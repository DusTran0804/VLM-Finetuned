```
import subprocess
import time
import urllib.request

# Khởi động FastAPI server ở chế độ nền
subprocess.Popen(["uvicorn", "server:app", "--host", "0.0.0.0", "--port", "8000"])
print("⏳ Đang khởi động Server...")
time.sleep(5)

# Khởi động localtunnel để lấy URL
!nohup lt --port 8000 > lt.log 2>&1 &
time.sleep(3)

# Lấy IP để xác minh bảo mật của localtunnel
ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip("\n")

print("\n" + "="*50)
print(f"🔑 MẬT KHẨU IP CỦA BẠN LÀ: {ip}")
print("="*50 + "\n")
print("🌐 ĐƯỜNG LINK PUBLIC URL CỦA BẠN LÀ:")
!cat lt.log
print("\n👉 HƯỚNG DẪN:")
print("1. Bấm vào link trên.")
print("2. Dán Mật khẩu IP vào ô 'Endpoint IP' và bấm Submit.")
print("3. Copy cái link đó và dán vào biến API_URL trên phần cài đặt Environment của lpr-client trên Render.")
```
